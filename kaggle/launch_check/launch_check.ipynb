```python
import json
from pathlib import Path
import torch

gpu_names = [torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())]
report = {"cuda_available": bool(torch.cuda.is_available()), "gpu_count": len(gpu_names), "gpu_names": gpu_names, "torch_version": torch.__version__, "purpose": "GitHub Actions Kaggle dual-T4 launch verification"}
destination = Path('/kaggle/working/t4_launch_check.json')
destination.write_text(json.dumps(report, indent=2) + '\n', encoding='utf-8')
print(json.dumps(report, indent=2))
assert report['cuda_available'] and report['gpu_count'] == 2 and all('T4' in name for name in gpu_names), 'Expected two NVIDIA T4 GPUs'
```
